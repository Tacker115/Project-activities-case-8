```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Настройки графиков
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (12, 5)

df = pd.read_csv('../data/processed/SeoulBikeData_clean.csv')

# ТЗ: Из Date сделать дополнительные признаки
df['Date'] = pd.to_datetime(df['Date'])
df['month'] = df['Date'].dt.month
df['day_of_week'] = df['Date'].dt.day_of_week # 0 = понедельник, 6 = воскресенье

# ТЗ: is_weekend (1 если сб или вс, иначе 0)
df['is_weekend'] = df['day_of_week'].apply(lambda x: 1 if x >= 5 else 0)

# Для удобства графиков переведем день недели в текст
days_map = {0: 'Mon', 1: 'Tue', 2: 'Wed', 3: 'Thu', 4: 'Fri', 5: 'Sat', 6: 'Sun'}
df['day_name'] = df['day_of_week'].map(days_map)

df.head(2)
```

```python
plt.figure(figsize=(8, 4))
sns.lineplot(data=df, x='Hour', y='Rented Bike Count', marker='o', color='blue', errorbar=None)
plt.title('Средний спрос на велосипеды по часам суток')
plt.xlabel('Час дня')
plt.ylabel('Среднее кол-во аренд')
plt.xticks(range(0, 24))
plt.grid(True)
plt.show()
```

## Вывод: 
### Спрос имеет два ярко выраженных пика — в 8:00 и в 18:00.<br>Это связано с поездками людей на работу и обратно.

```python
# Переименуем для красоты легенды
df['Тип дня'] = df['is_weekend'].map({0: 'Будни', 1: 'Выходные'})

plt.figure(figsize=(8, 4))
sns.lineplot(data=df, x='Hour', y='Rented Bike Count', hue='Тип дня', 
             palette={'Будни': 'red', 'Выходные': 'green'}, errorbar=None)
plt.title('Спрос по часам: Будни vs Выходные')
plt.xlabel('Час дня')
plt.ylabel('Среднее кол-во аренд')
plt.xticks(range(0, 24))
plt.show()
```

## Вывод: 
### В будни пики приходятся на 8:00 и 18:00.<br>В выходные спрос плавно возрастает к середине дня (12:00 - 16:00).

```python
plt.figure(figsize=(6, 3))
season_order = ['Winter', 'Spring', 'Summer', 'Autumn']

# Добавили hue='Seasons' и legend=False
sns.barplot(data=df, x='Seasons', y='Rented Bike Count', order=season_order, 
            palette='viridis', hue='Seasons', legend=False, errorbar=None)

plt.title('Средний спрос на велосипеды по сезонам')
plt.xlabel('Сезон')
plt.ylabel('Среднее кол-во аренд')
plt.show()
```

## Вывод: 
### Летом и осенью спрос максимален.<br>Зимой он самый низкий из-за холода.


```python
plt.figure(figsize=(7, 4))

# Добавили hue='Holiday' и legend=False
sns.barplot(data=df, x='Holiday', y='Rented Bike Count', 
            palette='Set2', hue='Holiday', legend=False, errorbar=None)

plt.title('Спрос: Праздники vs Обычные дни')
plt.xlabel('Тип дня')
plt.ylabel('Среднее кол-во аренд')
plt.show()
```

## Вывод: 
### В обычные дни спрос выше, чем в праздники, так как в праздники люди реже совершают рутинные поездки.


```python
pivot_table = df.pivot_table(index='day_name', columns='Hour', values='Rented Bike Count', aggfunc='mean')
days_order = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
pivot_table = pivot_table.reindex(days_order)

plt.figure(figsize=(14, 5))
sns.heatmap(pivot_table, cmap='YlOrRd', annot=False)
plt.title('Тепловая карта спроса: День недели и Час')
plt.xlabel('Час дня')
plt.ylabel('День недели')
plt.show()
```

## Вывод: 
### Видно, что пики в будни приходятся на 8 и 18 часов, а в выходные активность смещена на период с 12 до 16 часов.
