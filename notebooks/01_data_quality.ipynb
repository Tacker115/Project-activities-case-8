```python
import pandas as pd

df = pd.read_csv("../data/raw/SeoulBikeData.csv", encoding="unicode_escape")

print('Matrix:', df.shape)
print()
print('Columns:', df.columns)
print()
print('Data types:')
print(df.dtypes)
print()
print('Duplicates:', df.duplicated().sum())
print('NaN strings:', df.isna().sum().sum())
if df.isna().sum().sum() != 0:
    display(df[df.isna().any(axis=1)])

display(df.head())
display(df.describe())

print("Target min:", df["Rented Bike Count"].min())
print("Target max:", df["Rented Bike Count"].max())
```

```python
import pandas as pd
import os

raw_path = "../data/raw/SeoulBikeData.csv"
processed_path = "../data/processed/SeoulBikeData_clean.csv"

if os.path.exists(processed_path):
    print("Датасет уже преобразован: " + processed_path + ".")
else:
    print("Датасет ещё не преобразован...")
    df = pd.read_csv(raw_path, encoding="unicode_escape")
    df['Date'] = pd.to_datetime(df['Date'], format='%d/%m/%Y')
    os.makedirs(os.path.dirname(processed_path), exist_ok=True)
    df.to_csv(processed_path, index=False, encoding="utf-8")
    print("Данные обработаны и сохранены в: " + processed_path + ".")

print("Тип колонки Date:", df['Date'].dtype)
display(df.head(2))
```
